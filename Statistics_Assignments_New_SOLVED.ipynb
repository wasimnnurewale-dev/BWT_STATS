#### Q1. Arithmetic Mean:- The weekly sales (in units) of a retail store were: 48, 52, 61, 57, 49, 63, 70

#### Q8. Mode:- Determine the mode of the following dataset: 12, 15, 18, 15, 20, 18, 15, 22, 18, 15. State whether the distribution is unimodal, bimodal, or multimodal.

```python

from collections import Counter

data = [12, 15, 18, 15, 20, 18, 15, 22, 18, 15]
counts = Counter(data)
max_freq = max(counts.values())
modes = [val for val, freq in counts.items() if freq == max_freq]

print("Frequency counts:", dict(counts))
print("Mode(s):", modes, "with frequency", max_freq)
if len(modes) == 1:
    print("The distribution is unimodal.")
elif len(modes) == 2:
    print("The distribution is bimodal.")
else:
    print("The distribution is multimodal.")
```

```python

import numpy as np
from scipy.stats import gmean

growth = np.array([1.08, 1.12, 0.96, 1.10])
gm = gmean(growth)
gm_manual = (np.prod(growth)) ** (1 / len(growth))

print("Growth factors:", growth)
print("Geometric mean (scipy):", gm)
print("Geometric mean (manual check):", gm_manual)
print("=> Average annual growth factor is about %.4f, i.e. ~%.2f%% average growth." % (gm, (gm - 1) * 100))
```

#### Q22. Effect of Adding a Constant:- The observations are: 14, 18, 22, 26, 30. Without recalculating from scratch, determine the new:
> 1. Mean
> 2. Variance
> 3. Standard deviation

> if 8 is added to every observation.

```python

import numpy as np

data = np.array([14, 18, 22, 26, 30])
orig_mean, orig_var, orig_std = data.mean(), data.var(ddof=0), data.std(ddof=0)

# Adding a constant shifts the mean but leaves variance/std unchanged
new_mean = orig_mean + 8
new_var = orig_var
new_std = orig_std

print("Original -> Mean:", orig_mean, " Variance:", orig_var, " Std:", orig_std)
print("After adding 8 to every observation:")
print("New Mean:", new_mean)
print("New Variance (unchanged):", new_var)
print("New Standard Deviation (unchanged):", new_std)

# Verify directly
shifted = data + 8
print("Verification -> Mean:", shifted.mean(), " Variance:", shifted.var(ddof=0), " Std:", shifted.std(ddof=0))
```

#### Q23. Effect of Multiplying by a Constant:- The observations are: 5, 8, 11, 14, 17. Each observation is multiplied by 4. Determine the new:
> 1. Mean
> 2. Variance
> 3. Standard deviation

```python

import numpy as np

data = np.array([5, 8, 11, 14, 17])
orig_mean, orig_var, orig_std = data.mean(), data.var(ddof=0), data.std(ddof=0)

k = 4
# Multiplying by constant k: mean -> k*mean, variance -> k^2*var, std -> k*std
new_mean = k * orig_mean
new_var = (k ** 2) * orig_var
new_std = k * orig_std

print("Original -> Mean:", orig_mean, " Variance:", orig_var, " Std:", orig_std)
print("After multiplying every observation by 4:")
print("New Mean:", new_mean)
print("New Variance:", new_var)
print("New Standard Deviation:", new_std)

scaled = data * k
print("Verification -> Mean:", scaled.mean(), " Variance:", scaled.var(ddof=0), " Std:", scaled.std(ddof=0))
```

#### Q32. Two Coin Tosses:- Two fair coins are tossed simultaneously. Find the probability of obtaining:
> 1. Exactly one Head.
> 2. Two Heads.
> 3. At least one Head.

```python

from itertools import product

outcomes = list(product(["H", "T"], repeat=2))
total = len(outcomes)

exactly_one_head = sum(1 for o in outcomes if o.count("H") == 1) / total
two_heads = sum(1 for o in outcomes if o.count("H") == 2) / total
at_least_one_head = sum(1 for o in outcomes if o.count("H") >= 1) / total

print("Sample space:", outcomes)
print("P(exactly one Head):", exactly_one_head)
print("P(two Heads):", two_heads)
print("P(at least one Head):", at_least_one_head)
```

#### Q33. Three Coin Tosses:- Three fair coins are tossed. Find the probability of obtaining:
> 1. Exactly two Heads.
> 2. At most one Head.
> 3. No Heads.

```python

from itertools import product

outcomes = list(product(["H", "T"], repeat=3))
total = len(outcomes)

exactly_two = sum(1 for o in outcomes if o.count("H") == 2) / total
at_most_one = sum(1 for o in outcomes if o.count("H") <= 1) / total
no_heads = sum(1 for o in outcomes if o.count("H") == 0) / total

print("Total outcomes:", total)
print("P(exactly two Heads):", exactly_two)
print("P(at most one Head):", at_most_one)
print("P(no Heads):", no_heads)
```

#### Q34. Rolling a Die:- A fair die is rolled once. Find the probability of obtaining:

#### Q39. A box contains:

> - 4 Green balls

> Two balls are drawn without replacement. Find the probability that:

> 2. One is Red and the other is Blue.
> 3. At least one is Green.

```python

from math import comb

red, blue, green = 6, 5, 4
total = red + blue + green

p_both_red = comb(red, 2) / comb(total, 2)
p_one_red_one_blue = (red * blue) / comb(total, 2)
p_at_least_one_green = 1 - comb(total - green, 2) / comb(total, 2)

print("P(both Red):", p_both_red)
print("P(one Red and one Blue):", p_one_red_one_blue)
print("P(at least one Green):", p_at_least_one_green)
```

### Bayes' Theorem:- 

#### Q47. A disease affects 2% of a population. A diagnostic test has:
> - Sensitivity = 95%
> - Specificity = 98%

> A person tests positive. Find the probability that the person actually has the disease. 

```python

p_disease = 0.02
sensitivity = 0.95         # P(positive | disease)
specificity = 0.98         # P(negative | no disease)
p_false_positive = 1 - specificity   # P(positive | no disease)

p_no_disease = 1 - p_disease

p_positive = sensitivity * p_disease + p_false_positive * p_no_disease
p_disease_given_positive = (sensitivity * p_disease) / p_positive

print("P(test positive):", p_positive)
print("P(disease | positive) [Bayes' theorem]:", p_disease_given_positive)
```

#### Q96. A manufacturer claims that the average weight of a packet is 500 g. A sample of 64 packets has:
> - Mean = 495 g
> - Population standard deviation = 20 g

> At the 5% significance level:
> 1. State the null and alternative hypotheses.
> 2. Compute the Z statistic.
> 3. State the decision.
> 4. Draw the conclusion.

```python

import math
from scipy.stats import norm

mu0 = 500          # claimed mean
n, xbar, sigma = 64, 495, 20
alpha = 0.05

# H0: mu = 500   H1: mu != 500  (two-tailed)
SE = sigma / math.sqrt(n)
Z = (xbar - mu0) / SE
z_crit = norm.ppf(1 - alpha/2)
p_value = 2 * (1 - norm.cdf(abs(Z)))

print("H0: mu = 500 g   |   H1: mu != 500 g")
print("Z statistic:", Z)
print("Critical Z value (two-tailed, alpha=0.05):", z_crit)
print("p-value:", p_value)

if abs(Z) > z_crit:
    print("Decision: Reject H0 (|Z| > z_critical)")
    print("Conclusion: There is sufficient evidence that the average packet weight differs from 500 g.")
else:
    print("Decision: Fail to reject H0")
    print("Conclusion: There is not sufficient evidence that the average packet weight differs from 500 g.")
```

#### Q100. A machine is designed to fill bottles with 1 litre of liquid.
> A sample of 100 bottles has:
> - Mean = 0.985 litre
> - Population standard deviation = 0.04 litre

> At the 5% significance level:
> 1. State the null and alternative hypotheses.
> 2. Compute the Z statistic.
> 3. Determine the p-value.
> 4. State whether the null hypothesis should be rejected.
> 5. Construct a 95% confidence interval for the population mean.
> 6. Interpret the confidence interval in the context of the problem.

```python

import math
from scipy.stats import norm

mu0 = 1.0
n, xbar, sigma = 100, 0.985, 0.04
alpha = 0.05

# H0: mu = 1 litre   H1: mu != 1 litre  (two-tailed)
SE = sigma / math.sqrt(n)
Z = (xbar - mu0) / SE
z_crit = norm.ppf(1 - alpha/2)
p_value = 2 * (1 - norm.cdf(abs(Z)))

print("H0: mu = 1.0 litre   |   H1: mu != 1.0 litre")
print("Z statistic:", Z)
print("Critical Z value (two-tailed, alpha=0.05):", z_crit)
print("p-value:", p_value)

if p_value < alpha:
    print("Decision: Reject H0 (p-value < alpha).")
else:
    print("Decision: Fail to reject H0 (p-value >= alpha).")

# 95% Confidence Interval
z_95 = norm.ppf(0.975)
margin = z_95 * SE
ci_lower, ci_upper = xbar - margin, xbar + margin
print(f"95% Confidence Interval for the population mean: ({ci_lower:.5f}, {ci_upper:.5f}) litres")

print()
print("Interpretation: We are 95% confident that the true mean fill volume lies within this interval.")
print("Since 1.0 litre is NOT contained in the interval, this confirms the machine is under-filling")
print("bottles on average, consistent with rejecting H0.")
```
